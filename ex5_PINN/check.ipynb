```python
import importlib

def check_package(name, required_version=None, alt_name=None):
    """Checks if a package is installed and matches expected version."""
    try:
        module = importlib.import_module(name if not alt_name else alt_name)
        version = getattr(module, '__version__', 'unknown')
        if required_version and version != required_version:
            print(f"[❗] {name}: Version {version} (Expected: {required_version})")
        else:
            print(f"[✅] {name}: Version {version}")
    except ImportError:
        print(f"[❌] {name} is NOT installed")

print("🔍 Checking key packages in current environment...\n")

# Core packages
check_package("tensorflow", "2.3.1")
check_package("keras", "2.4.3")
check_package("sciann", "0.5.1")
check_package("numpy", "1.18.5")
check_package("matplotlib", "3.1.3")
check_package("sympy", "1.4")
check_package("sklearn", alt_name="scikit_learn")

# Optional but useful
check_package("torch")  # Used in your PINN notebook
check_package("ipykernel")
check_package("protobuf", "3.20.3")

print("\n✅ Done checking environment.")
```
